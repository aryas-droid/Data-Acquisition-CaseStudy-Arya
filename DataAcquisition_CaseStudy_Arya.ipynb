TASK-6
Rename columns wherever necessary so that all datasets have a common column
name for merging.
Hint:
● Example: Country → country

```python
print("Population")
c=pop.rename(columns={"country":"country"})
c
```

```python
print("GDP")
d=gd.rename(columns={"Country":"country"})
d
```

```python
r=iu.rename(columns={"Country":"country"})
r
```

```python
s=lr.rename(columns={"name":"country"})
s
```

```python
# Convert required columns to numeric data types
# Convert required columns to numeric data types
pop["population"] = pd.to_numeric(pop["population"], errors="coerce")
gd["GDP"] = pd.to_numeric(gd["GDP"], errors="coerce")
iu["internet_users"] = pd.to_numeric(iu["internet_users"], errors="coerce")
lr["literacy_rate"] = pd.to_numeric(lr["literacy_rate"], errors="coerce")

print(pop.dtypes)
print(gd.dtypes)
print(iu.dtypes)
print(lr.dtypes)
```

# Task 8
Merge all four datasets into one DataFrame.
Hint:
● Use the common column (country)
● Choose an appropriate join method.

```python
merged_pop=c.merge(d,on="country",how="inner")
merged_pop=merged_pop.merge(r,on="country",how="inner")
merged_pop=merged_pop.merge(s,on="country",how="inner")
print(merged_pop)
```

#Task 9
Create the following calculated columns:
1. Internet Penetration (%)
2. GDP Per Capita
Hint:
Round the calculated values to two decimal places.


```python
# Task 9: Create calculated columns

# Internet Penetration (%)
merged_pop["Internet Penetration (%)"] = (
    merged_pop["internet_users"] / merged_pop["population"] * 100
).round(2)

# GDP Per Capita
merged_pop["GDP Per Capita"] = (
    merged_pop["GDP"] / merged_pop["population"]
).round(2)

# Display the result
print(merged_pop)
```

#Part D – Load Data
#Task 10
Create a SQLite database. Store the final merged dataset into a table named
‘country_statistics’
Hint:
● Use the sqlite3 library.
● Load the DataFrame using pandas.

```python
import sqlite3

# Create SQLite database
conn = sqlite3.connect("country_statistics.db")

# Store the final merged DataFrame into SQLite
merged_pop.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)

print("Data successfully stored in SQLite database.")
```

#Task 11
Verify that the data has been loaded successfully.
Hint:
Retrieve the first five records from the SQLite table.


```python
# Retrieve first five records from SQLite table

result = pd.read_sql_query(
    "SELECT * FROM country_statistics LIMIT 5",
    conn
)

print(result)
```

#Answer the following questions using pandas.
1 List the Top 5 countries by GDP.
2 List the Top 5 countries by population.
3 Calculate the average literacy rate.
4 List all countries having a literacy rate greater than 90%.
5 List all countries having an internet penetration greater than 70%.
6 List the Top 5 countries by GDP per capita.
7 Calculate the total population represented in the dataset.
8 Calculate the average internet penetration across all countries.
9 Identify the country having the highest number of internet users.
10 Find the correlation between Literacy Rate and Internet Penetration.
Hint:
Use the appropriate pandas statistical function.


```python
# 1. Top 5 countries by GDP
print("1. Top 5 countries by GDP:")
print(merged_pop.nlargest(5, "GDP")[["country", "GDP"]])
print("___________________________________________________________________")
# 2.Top 5 countries by population
print("\n2. Top 5 countries by population:")
print(merged_pop.nlargest(5,"population")[["country","population"]])
print("___________________________________________________________________")
print("\n3. Calculate the average literacy rate")
print(merged_pop["literacy_rate"].mean())
print("___________________________________________________________________")
print("\n4. List all countries having a literacy rate greater than 90%")
print(merged_pop[merged_pop["literacy_rate"] >90])
print("___________________________________________________________________")

```

```python
print("\n5. List all countries having an internet penetration greater than 70%")
print(merged_pop[merged_pop["Internet Penetration (%)"] >70])
print("___________________________________________________________________")
```

```python
print("\n6. List the Top 5 countries by GDP per capita")
print(merged_pop.nlargest(5,"GDP Per Capita")[["country","GDP Per Capita"]])
print("___________________________________________________________________")
```

```python
print("\n7. Calculate the total population represented in the dataset")
print(merged_pop["population"].sum())
print("___________________________________________________________________")
```

```python
print("\n8. Calculate the average internet penetration across all countries")
print(merged_pop["Internet Penetration (%)"].mean())
print("___________________________________________________________________")
print("\n9. Identify the country having the highest number of internet users")
print(merged_pop[merged_pop["internet_users"]==merged_pop["internet_users"].max()][["country","internet_users"]])
print("___________________________________________________________________")
print("\n10. Find the correlation between Literacy Rate and Internet Penetration")
print(merged_pop["literacy_rate"].corr(merged_pop["Internet Penetration (%)"]))
```
